```python
def maxcut_cost_hamiltonian(num_nodes, edges):
    """Build the Max-Cut cost Hamiltonian sum_(i,j) 0.5 * Z_i Z_j.
    You should return a SparsePauliOp containing all of the terms of the hamiltonian"""
    return SparsePauliOp.from_sparse_list([("ZZ", [e[0], e[1]], 0.5) for e in edges], num_qubits=num_nodes)
```
