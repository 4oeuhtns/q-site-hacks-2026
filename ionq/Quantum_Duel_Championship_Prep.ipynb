# Championship Submission Check

**Check what your defender actually submitted, and what that correction achieved.**

This optional notebook helps you inspect a submission for the **eight-qubit open architecture final**. Bring the `submission.zip` you plan to upload. You can review saved local test results or run the existing local tests from here.

The central question is:

**Did the program run? → Was its correction legal? → Which correction was recorded at each checkpoint? → Did that correction recover the process?**

These are separate checks. A program can finish successfully and earn zero recovery points. It can also estimate the hidden operation well but fail to turn that estimate into a useful, legal correction in time for a checkpoint.

**Nothing about the competition changes.** The rules, gate limits, shot budget, scoring, runtime allocation, deadline, and submission format stay the same. This notebook is optional, provides no new solver, and requires no additional deliverable. Continue developing in your usual workflow and upload your complete ZIP through the existing competition portal.


## Understand the four stages

Throughout this notebook, a **patch** means the correction circuit submitted by your defender. A **checkpoint** records the patch in effect at that stage. **Evidence** means the saved JSON results of a local test.

| Stage | What needs to happen | What this notebook can show |
|---|---|---|
| **Execution** | Your program runs and completes its interaction with the local runner. | Completion status or a recorded failure. |
| **Inference** | Your defender uses measurements to estimate the hidden operation. | It cannot inspect or judge that internal estimate. Use your own diagnostics. |
| **Synthesis** | Your defender turns its estimate into a correction circuit within the allowed gates and limits. | It checks the legality and size of recorded patches, but cannot see candidates that were never submitted. |
| **Submitted correction** | A useful patch reaches the runner before the checkpoint closes. | It summarizes the patch recorded at each checkpoint and its local recovery error and points. |

**Execution ≠ inference ≠ synthesis ≠ submitted correction.** A success at one stage does not establish success at the next.

For example, a defender might construct a promising estimate, fail to compile a usable correction, and finish with an empty patch. The local worker can report `PASSED` because execution completed. The recorded patch can be legal because an empty circuit is allowed. Recovery can still earn zero points. The report reveals that outcome; your own logs are needed to establish its cause.

## 2. Check the final profile and recovery objective

The final uses **`quantum-duel-8q-open-0.7.1`**. The cells read its limits from the included public SDK; the table below shows the values used for these local checks. The portal's frozen round remains authoritative.

### What “open architecture” means

Cup 2 promised a particular frame structure. The final does not. Legal attacks can use chronological sequences of `rx`, `ry`, `rz`, `rxx`, `ryy`, and `rzz`, with repeated pairs, arbitrary ordering, continuous angles, and shorter circuits, subject to the published admission rules and limits. Six layers, a fixed matching, and a small number of non-Clifford rotations are not guaranteed.

Each submitted attack template fixes its gate sequence and targets; its declared parameters vary between realizations. A defender designed around one structural assumption may work on some legal attacks and fail on others. Inference, synthesis, and runtime can all behave differently across those structures.

### What recovery measures

Let $A$ be the hidden attack and $P$ the recorded correction applied after it. The goal is for $PA$ to act like identity, up to a global phase. For eight qubits, the process error is

$$
\epsilon(P,A)=1-\frac{|\operatorname{Tr}(PA)|^2}{65,536}.
$$

**Lower error is better.** Each checkpoint earns 100 points at error at most `0.001`, zero points at error at least `0.1`, and a logarithmically scaled score between those thresholds. The three checkpoint scores count equally.

Recovery points are not a fidelity percentage: for example, error `0.01` earns 50 points. A small process error also does not guarantee the same accuracy on every possible input. This notebook displays recorded local errors and checks the score arithmetic; it does not independently recompute the hidden process from saved evidence.

### Load the review helpers and check the package

Run the folded helper cell, then the package-check cell below it. Defining these helpers does not execute your defender.

````python
"""Read-only display helpers for existing Quantum Duel local smoke results.

No solver execution, new observations, compilation, upload, or private score access.
The notebook's existing validator remains the authority for local admission checks.
"""
def summarize_local_diagnostics(evidence, archive, rules):
    import hashlib
    import math
    from pathlib import Path
    from duelkit.quantum import from_data, validate
    from qduel_sdk.serialization import digest

    if not isinstance(evidence, dict) or not isinstance(evidence.get("result"), dict):
        raise ValueError("Pass the existing SMOKE_EVIDENCE, not a competition replay.")
    result = evidence["result"]
    cases = result.get("cases")
    if not isinstance(cases, list):
        raise ValueError("The local report has no cases list.")
    current_rules = rules.model_dump()
    expected_rules = digest(current_rules)
    try:
        current_archive = hashlib.sha256(Path(archive).read_bytes()).hexdigest()
        archive_state = ("MATCH" if evidence.get("archive_sha256") == current_archive
                         else "CHANGED OR UNBOUND")
    except OSError:
        current_archive = None
        archive_state = "FILE UNAVAILABLE"
    rules_state = "MATCH" if evidence.get("rules_sha256") == expected_rules else "CHANGED OR UNBOUND"
    report = {"mode": "LOCAL_DIAGNOSTIC_ONLY", "profile": rules.version,
              "archive_sha256": current_archive, "archive_binding": archive_state,
              "rules_binding": rules_state, "cases": [], "checkpoints": [], "notes": []}
    if archive_state != "MATCH" or rules_state != "MATCH":
        report["notes"].append("STALE OR UNBOUND: these results do not describe the selected ZIP and rules. Rebuild and rerun the existing smoke tests.")
    report["notes"].append("Matching hashes refer to packaged bytes, not unsaved edits in main.py. Rebuild the ZIP after changing source.")

    def finite_number(value):
        return type(value) in (int, float) and math.isfinite(value)

    def budget(value, limit):
        if type(value) is not int or value < 0:
            return "not recorded", "Budget use is missing or malformed."
        return f"{value:,} / {limit:,}", ("Recorded budget exceeds the selected limit." if value > limit else "")

    for number, row in enumerate(cases, 1):
        if not isinstance(row, dict):
            report["cases"].append({"case": f"Case {number}", "execution": "Unknown",
                                   "recovery": "Unavailable", "notes": "Malformed local record."})
            continue
        label = str(row.get("case", f"Case {number}"))
        if row.get("seed") is not None:
            label += f" (seed {row['seed']})"
        if row.get("mode") not in (None, "LOCAL_PRACTICE_NOT_OFFICIAL"):
            raise ValueError("This helper accepts local practice reports only, not live or official results.")
        issues = []
        profile_ok = row.get("rules") == current_rules
        raw_checkpoints = row.get("checkpoint_scores", [])
        if not isinstance(raw_checkpoints, list):
            raw_checkpoints = []
            issues.append("Malformed checkpoint list.")
        valid_points = []
        stages = []
        empty_stages = []
        local_checkpoint_rows = []
        patches_valid = True
        for cp in raw_checkpoints:
            if not isinstance(cp, dict):
                issues.append("Malformed checkpoint record.")
                patches_valid = False
                continue
            stage = cp.get("stage")
            stage_ok = type(stage) is int and 1 <= stage <= rules.checkpoints
            stages.append(stage if stage_ok else -1)
            patch = cp.get("patch")
            legal = "Not checked"
            gates = entanglers = "not recorded"
            if profile_ok and isinstance(patch, list):
                try:
                    parsed = from_data(patch)
                    validate(parsed, **rules.validation_kwargs("patch"))
                    gates = f"{len(parsed)} / {rules.patch_max_gates}"
                    entanglers = f"{sum(len(g.targets) == 2 for g in parsed)} / {rules.patch_max_entanglers}"
                    legal = "Legal"
                    if not parsed:
                        empty_stages.append(stage)
                        legal = "Legal identity / empty patch"
                except (ValueError, TypeError, KeyError, IndexError) as exc:
                    legal = "INVALID RECORD: " + str(exc)[:180]
                    patches_valid = False
            else:
                patches_valid = False
                legal = "Not checked: missing patch or different rules"
            error, points = cp.get("process_infidelity"), cp.get("recovery_points")
            score_ok = (profile_ok and stage_ok and finite_number(error) and 0 <= error <= 1
                        and finite_number(points) and 0 <= points <= 100)
            if score_ok:
                expected = (100.0 if error <= rules.good_error else 0.0 if error >= rules.bad_error
                            else 100 * math.log(rules.bad_error / error) / math.log(rules.bad_error / rules.good_error))
                score_ok = math.isclose(points, expected, abs_tol=1e-7, rel_tol=1e-9)
            if score_ok:
                valid_points.append(float(points))
            else:
                issues.append("Missing, mismatched, or inconsistent checkpoint score; do not interpret it as zero.")
            local_checkpoint_rows.append({"case": label, "checkpoint": stage,
                "shots": cp.get("spent_shots", "not recorded"), "gates": gates,
                "entanglers": entanglers, "legality": legal,
                "error": f"{error:.6g}" if score_ok else "unavailable",
                "points": f"{points:.2f}" if score_ok else "unavailable"})
        complete = (profile_ok and len(stages) == rules.checkpoints
                    and sorted(stages) == list(range(1, rules.checkpoints + 1))
                    and len(valid_points) == rules.checkpoints and patches_valid)
        if raw_checkpoints and not complete:
            issues.append("Incomplete or inconsistent evidence; no complete recovery verdict.")
        if not profile_ok and raw_checkpoints:
            issues.append("Recorded case rules do not match the selected profile.")
        if empty_stages:
            issues.append("Identity / empty correction was recorded at checkpoints " + ", ".join(map(str, empty_stages)) + ". This does not identify why the solver chose it.")
        execution = ("Completed" if row.get("status") == "PASSED" else
                     "Failed" if row.get("status") == "FAILED" else "Not established")
        recovery = "Not recorded / not established"
        if complete:
            mean = sum(valid_points) / len(valid_points)
            recovery = ("Full credit on this local case" if all(p >= 100 - 1e-7 for p in valid_points)
                        else "Completed checkpoints; ZERO recovery points" if all(p == 0 for p in valid_points)
                        else f"Local recovery: {mean:.2f} / 100")
            if finite_number(row.get("recovery_points")) and not math.isclose(row["recovery_points"], mean, abs_tol=1e-7):
                issues.append("Recorded encounter average disagrees with its checkpoint scores.")
                recovery = "Inconsistent local average; inspect the report"
        if execution == "Failed":
            issues.append("Recorded runtime error: " + str(row.get("error", "reason unavailable"))[:300])
            issues.append("No missing checkpoint is assigned a score by this display. Do not infer a timeout or compiler failure without evidence.")
        resources = {}
        for key, limit in (("spent_shots", rules.total_shots), ("distinct_settings", rules.max_settings), ("requests", rules.max_requests)):
            resources[key], warning = budget(row.get(key), limit)
            if warning:
                issues.append(key + ": " + warning)
        report["cases"].append({"case": label, "execution": execution, "recovery": recovery,
            "shots": resources["spent_shots"], "settings": resources["distinct_settings"],
            "actions": resources["requests"], "notes": " ".join(dict.fromkeys(issues))})
        report["checkpoints"].extend(local_checkpoint_rows)
    report["notes"].append("Read-only view of recorded local evidence: no new execution, shots, score access, or upload. This is neither an official result nor independent verification of process fidelity.")
    report["notes"].append("The public local and zz cases are simple teaching checks. Mixed, multilayer and frame are broader public examples, not exhaustive coverage or a prediction of championship performance.")
    report["notes"].append("Runtime duration and the solver's internal stage are not present in these reports. This panel does not estimate them or change timeout handling.")
    return report


def show_local_diagnostics(evidence, archive, rules):
    """Notebook renderer. Requires the existing, HTML-escaping show_rows helper."""
    report = summarize_local_diagnostics(evidence, archive, rules)
    print("LOCAL PRACTICE — execution, circuit validity, and recovery are different checks.")
    print("ZIP binding:", report["archive_binding"], "| Rules binding:", report["rules_binding"])
    print("Profile:", report["profile"])
    for note in report["notes"]:
        print("Note:", note)
    show_rows(report["cases"], [("case", "Local case"), ("execution", "Execution"),
        ("recovery", "Recorded recovery"), ("shots", "Shots"), ("settings", "Settings"),
        ("actions", "Actions"), ("notes", "Details")])
    if report["checkpoints"]:
        show_rows(report["checkpoints"], [("case", "Case"), ("checkpoint", "Checkpoint"),
            ("shots", "Actual shots used"), ("gates", "Gates / cap"),
            ("entanglers", "Entanglers / cap"), ("legality", "Patch check"),
            ("error", "Recorded process infidelity"), ("points", "Recovery points (not fidelity %)")])
    return report




def _finite_int(value, minimum=0):
    return type(value) is int and value >= minimum


def _read_json(path, limit=32_000_000):
    import json
    from pathlib import Path
    path = Path(path).expanduser()
    with path.open("rb") as stream:
        raw = stream.read(limit + 1)
    if len(raw) > limit:
        raise ValueError("Local JSON is too large for this optional viewer.")
    def unique(pairs):
        out = {}
        for key, value in pairs:
            if key in out:
                raise ValueError("Duplicate JSON key: " + str(key))
            out[key] = value
        return out
    def bad_number(value):
        raise ValueError("Nonfinite JSON number: " + value)
    return json.loads(raw, object_pairs_hook=unique, parse_constant=bad_number)


def inspect_candidate(path, rules):
    """Run the unchanged nonexecuting validator. No extraction or participant imports."""
    import hashlib
    from pathlib import Path
    from qduel_sdk.archive import MAX_ARCHIVE
    from qduel_sdk.submission import validate_submission_archive
    path = Path(path).expanduser()
    try:
        with path.open("rb") as stream:
            blob = stream.read(MAX_ARCHIVE + 1)
        checked = validate_submission_archive(blob, rules)
    except (OSError, ValueError, TypeError, KeyError) as exc:
        return {"status": "NOT_VALIDATED", "valid": False, "path": str(path),
                "error": f"{type(exc).__name__}: {exc}", "code_execution": "NOT_RUN"}
    return {"status": "STATIC_VALIDATED", "valid": True, "path": str(path),
            "sha256": hashlib.sha256(blob).hexdigest(), "checked": checked,
            "code_execution": "NOT_RUN"}


def load_saved_evidence(path):
    """Recognize existing local formats without creating missing provenance."""
    import copy
    data = _read_json(path)
    if isinstance(data, dict) and isinstance(data.get("result"), dict) and isinstance(data["result"].get("cases"), list):
        return copy.deepcopy(data)
    if isinstance(data, dict) and isinstance(data.get("smoke"), dict):
        # Existing local_validation_report.json may lack rules_sha256. Leave it
        # missing; do not turn current rules into evidence about a previous run.
        return {"archive_sha256": data.get("archive_sha256"),
                "rules_sha256": data.get("rules_sha256"),
                "result": copy.deepcopy(data["smoke"]),
                "origin": "Existing local validation report; missing hashes stay missing"}
    if isinstance(data, dict) and isinstance(data.get("cases"), list):
        return {"archive_sha256": None, "rules_sha256": None,
                "result": copy.deepcopy(data), "origin": "Unbound raw local smoke output"}
    if isinstance(data, list) and all(isinstance(row, dict) and "slot" in row for row in data):
        rows = copy.deepcopy(data)
        for row in rows:
            row.setdefault("case", "Own attack slot " + str(row["slot"]))
        return {"archive_sha256": None, "rules_sha256": None,
                "result": {"cases": rows}, "origin": "Unbound old own-attack output"}
    raise ValueError("Choose local smoke_evidence.json, local_validation_report.json, or own_attack_results.json. Official replays are not supported here.")


def evidence_assessment(evidence, archive, rules, preflight=None):
    """Summarize provenance and evidence quality, never certify competitiveness."""
    from qduel_sdk.serialization import digest
    report = summarize_local_diagnostics(evidence, archive, rules)
    bound = report["archive_binding"] == "MATCH" and report["rules_binding"] == "MATCH"
    report["evidence_scope"] = ("SELECTED ZIP / SELECTED FINAL PROFILE" if bound
                                else "HISTORICAL OR UNBOUND — NOT EVIDENCE FOR THE SELECTED ZIP")
    warnings = []
    rows = evidence["result"]["cases"]
    if not rows:
        warnings.append("No local cases are recorded.")
    seen = set()
    for i, row in enumerate(rows, 1):
        if not isinstance(row, dict):
            warnings.append(f"Case {i}: malformed record.")
            continue
        label = str(row.get("case", f"Case {i}"))
        key = (label, str(row.get("seed", "not recorded")))
        if key in seen:
            warnings.append(f"{label}: duplicate case/seed; do not count it as extra evidence.")
        seen.add(key)
        if row.get("rules") != rules.model_dump():
            warnings.append(f"{label}: missing or different profile metadata.")
        if row.get("status") not in ("PASSED", "FAILED"):
            warnings.append(f"{label}: execution status is not established.")
        for field, cap in (("spent_shots", rules.total_shots),
                           ("distinct_settings", rules.max_settings),
                           ("requests", rules.max_requests)):
            value = row.get(field)
            if not _finite_int(value) or value > cap:
                warnings.append(f"{label}: {field} is missing, malformed or over the limit.")
        cps = row.get("checkpoint_scores", [])
        previous_shots = 0
        if isinstance(cps, list):
            for cp in cps:
                if not isinstance(cp, dict):
                    continue
                spent, stage = cp.get("spent_shots"), cp.get("stage")
                if (not _finite_int(spent) or not _finite_int(stage, 1)
                    or stage > rules.checkpoints or spent < previous_shots
                    or spent > stage * rules.block):
                    warnings.append(f"{label}: checkpoint shot accounting is missing or inconsistent.")
                elif _finite_int(row.get("spent_shots")) and spent > row["spent_shots"]:
                    warnings.append(f"{label}: checkpoint shots exceed the recorded encounter total.")
                if _finite_int(spent):
                    previous_shots = spent
                if "patch_hash" in cp and isinstance(cp.get("patch"), list):
                    if cp["patch_hash"] != digest(cp["patch"]):
                        warnings.append(f"{label}: checkpoint patch checksum disagrees with the gate list.")
    if not bound:
        warnings.insert(0, "Results are not bound to this ZIP and final profile; do not treat historical passes as current readiness.")
    if preflight is not None:
        if not preflight.get("valid") or preflight.get("sha256") != report["archive_sha256"]:
            warnings.insert(0, "The selected file did not pass the current static check, or changed after it.")
    report["review_warnings"] = list(dict.fromkeys(warnings))
    return report


def print_assessment(evidence, archive, rules, preflight=None):
    report = evidence_assessment(evidence, archive, rules, preflight)
    print("LOCAL READINESS REVIEW —", report["evidence_scope"])
    print("Execution, circuit legality and recovery are separate checks.")
    print("ZIP:", report["archive_binding"], "| Rules:", report["rules_binding"])
    for warning in report["review_warnings"]:
        print("CHECK:", warning)
    show_rows(report["cases"], [("case", "Case"), ("execution", "Execution"),
        ("recovery", "Recorded recovery"), ("shots", "Shots"), ("settings", "Settings"),
        ("actions", "Actions"), ("notes", "Recorded details")])
    if report["checkpoints"]:
        show_rows(report["checkpoints"], [("case", "Case"), ("checkpoint", "Checkpoint"),
            ("shots", "Shots used"), ("gates", "Gates / limit"), ("entanglers", "Entanglers / limit"),
            ("legality", "Circuit check"), ("error", "Recorded process infidelity"),
            ("points", "Points, not fidelity %")])
    for note in report["notes"]:
        print("Note:", note)
    return report


def _check_run_options(cases, seeds, timeout, rules):
    from qduel_sdk.profiles import practice_bank
    if not isinstance(cases, (list, tuple)) or not cases or any(not isinstance(c, str) for c in cases):
        raise ValueError("Choose a nonempty list of public case names.")
    if len(set(cases)) != len(cases):
        raise ValueError("Remove duplicate public case names.")
    unknown = set(cases) - set(practice_bank(rules))
    if unknown:
        raise ValueError("Unknown cases for this final: " + ", ".join(sorted(unknown)) +
                         ". Available: " + ", ".join(practice_bank(rules)))
    if (not isinstance(seeds, (tuple, list)) or not seeds
        or any(not _finite_int(s) or s > 2**32 - 100001 for s in seeds)
        or len(set(seeds)) != len(seeds)):
        raise ValueError("Use distinct nonnegative integer seeds below 2**32 - 100000.")
    if not _finite_int(timeout, 1):
        raise ValueError("Local timeout must be a positive integer in seconds.")


def _save_new_json(folder, prefix, data):
    import datetime, json, uuid
    from pathlib import Path
    folder = Path(folder).expanduser()
    folder.mkdir(parents=True, exist_ok=True)
    name = f"{prefix}_{datetime.datetime.now(datetime.timezone.utc).strftime('%Y%m%dT%H%M%SZ')}_{uuid.uuid4().hex[:8]}.json"
    path = folder / name
    # Exclusive creation: never overwrite old evidence or a participant artifact.
    with path.open("x", encoding="utf-8") as stream:
        json.dump(data, stream, indent=2, allow_nan=False)
    return path


def run_existing_public_tests(archive, rules, *, cases, seeds, timeout, trusted_code,
                              output_dir, maximum_runs=4):
    """Explicit opt-in use of the unchanged SDK smoke runner; no new solver."""
    import hashlib, tempfile
    from pathlib import Path
    from qduel_sdk.archive import MAX_ARCHIVE
    from qduel_sdk.serialization import digest
    from qduel_sdk.submission import smoke_submission, validate_submission_archive
    if trusted_code is not True:
        raise ValueError("Explicitly set TRUSTED_CODE=True only for your own or reviewed code.")
    _check_run_options(cases, seeds, timeout, rules)
    if not _finite_int(maximum_runs, 1) or len(cases)*len(seeds) > maximum_runs:
        raise ValueError("This local plan exceeds MAX_LOCAL_RUNS. Reduce it or explicitly increase that local safety guard.")
    path = Path(archive).expanduser()
    with path.open("rb") as stream:
        blob = stream.read(MAX_ARCHIVE + 1)
    validate_submission_archive(blob, rules)
    fingerprint = hashlib.sha256(blob).hexdigest()
    print(f"Explicit local execution: {len(cases)*len(seeds)} cases, up to {rules.total_shots:,} simulated shots per case.")
    print(f"The local timeout is {timeout}s PER CASE, not the official runtime budget.")
    # Run a fixed byte snapshot. Results bind what was actually run, even if the
    # original source ZIP changes while a test is running.
    with tempfile.TemporaryDirectory(prefix="qduel-readiness-input-") as tmp:
        fixed = Path(tmp) / "submission.zip"
        fixed.write_bytes(blob)
        result = smoke_submission(fixed, rules=rules, cases=tuple(cases), seeds=tuple(seeds),
                                  trusted_code=True, timeout=timeout)
    evidence = {"archive_sha256": fingerprint, "rules_sha256": digest(rules.model_dump()),
                "cases": list(cases), "seeds": list(seeds), "result": result,
                "origin": "Existing SDK local tests; explicitly requested from readiness notebook"}
    saved = _save_new_json(output_dir, "smoke_evidence", evidence)
    return evidence, saved
````

### Then: read the two tables together

| Table | What to read |
|---|---|
| **Case summary** | Execution status, recorded recovery, shots used, distinct settings, actions, and recorded failures or warnings. |
| **Checkpoint detail** | The correction's gate and entangler counts, legality, recorded process error, and recovery points at each checkpoint. |

Gate counts describe the **recorded correction**, not your attack template or an internal candidate circuit. Entanglers are the two-qubit gates counted against the published limit. Shots, settings, and actions describe recorded budget use; this report does not measure wall-clock duration, memory use, or time spent in inference versus synthesis.

### Which correction gets scored?

In the included local runner, submitting a patch replaces the current correction. Closing a checkpoint records the correction then in effect. A later patch does not change an earlier checkpoint. If no new patch is submitted, the current one remains in effect; before any submission, that patch is empty and acts as identity.

The local worker calls `finish()` after your program returns. That closes any remaining checkpoints with the current patch, so `PASSED` does not prove that a new, useful correction was submitted at every stage. The **Shots used** column records actual cumulative measurements; finishing early can produce checkpoint rows below the available shot budget.

An empty patch is legal and applies no correction. A missing checkpoint record is different: its outcome is unavailable, and this viewer does not assign it zero points. To inspect the full gate list, open the evidence JSON under `result → cases → checkpoint_scores → patch` for reports saved by this notebook.

These are local practice results. Use the competition portal for official leaderboard and replay results.

````python
import subprocess, tempfile, zipfile, io
import qduel_sdk
EMBEDDED_SDK = Path(qduel_sdk.__file__).resolve().parents[1]  # The public SDK directory written by Setup.

_OWN_WORKER = r"""
import importlib.util, inspect, json, sys
from pathlib import Path
from qduel_sdk.submission import validate_submission_archive
from qduel_sdk.archive import safe_extract
from qduel_sdk.rules import Rules
from qduel_sdk.contracts import Template
from qduel_sdk.templates import instantiate
from qduel_sdk.local import LocalSession
cfg = json.loads(sys.argv[1]); session = None
try:
    blob = Path(cfg['archive']).read_bytes()
    checked = validate_submission_archive(blob, Rules(**cfg['rules']))
    root = Path.cwd() / 'solution'; safe_extract(blob, root)
    rules = Rules(**checked['rules'])
    attack = instantiate(Template(**checked['attacks'][cfg['slot']]), cfg['seed'], rules)
    session = LocalSession(attack, rules, seed=cfg['seed'] + 100000)
    sys.path.insert(0, str(root))
    spec = importlib.util.spec_from_file_location('participant_main', root / 'main.py')
    module = importlib.util.module_from_spec(spec); sys.modules['participant_main'] = module
    spec.loader.exec_module(module)
    value = module.run(session.client(), rules)
    if inspect.isawaitable(value) or inspect.isgenerator(value):
        raise ValueError('run must execute synchronously')
    session.client().finish()
    result = dict(status='PASSED', **session.result())
except BaseException as exc:
    result = dict(status='FAILED', error=f'{type(exc).__name__}: {str(exc)[:400]}',
                  spent_shots=session.status()['spent_shots'] if session else 0)
Path(cfg['output']).write_text(json.dumps(result), encoding='utf-8')
"""


def test_own_attacks(archive, rules, seeds=(73,), *, trusted_code=False, timeout=900):
    if not trusted_code:
        raise ValueError('Set trusted_code=True only for your own or reviewed source.')
    preflight = validate_solution(archive, rules=rules)
    if not preflight['valid_for_upload']:
        raise ValueError(preflight['errors'])
    rows = []
    for slot in (0, 1):
        for seed in seeds:
            with tempfile.TemporaryDirectory(prefix='duel-own-') as directory:
                folder = Path(directory); result_file = folder / 'result.json'
                cfg = dict(archive=str(Path(archive).resolve()), rules=rules.model_dump(),
                           slot=slot, seed=int(seed), output=str(result_file))
                env = {**os.environ, 'PYTHONPATH': str(EMBEDDED_SDK), 'OPENBLAS_NUM_THREADS': '1', 'OMP_NUM_THREADS': '1'}
                with (folder / 'worker.log').open('wb') as stream:
                    try:
                        worker = subprocess.run([sys.executable, '-c', _OWN_WORKER, json.dumps(cfg)],
                            cwd=folder, env=env, stdout=stream, stderr=stream, timeout=timeout, check=False)
                        result = json.loads(result_file.read_text()) if result_file.exists() else {
                            'status': 'FAILED', 'error': f'Process returned {worker.returncode} without a result'}
                    except subprocess.TimeoutExpired:
                        result = {'status': 'FAILED', 'error': 'Local timeout; not a hostile-code containment guarantee'}
                rows.append(dict(slot=slot+1, seed=int(seed), **result))
    return rows



def run_existing_own_tests(archive, rules, *, seeds, timeout, trusted_code, output_dir,
                           maximum_runs=4):
    import hashlib, tempfile
    from pathlib import Path
    from qduel_sdk.archive import MAX_ARCHIVE
    from qduel_sdk.serialization import digest
    from qduel_sdk.submission import validate_submission_archive
    if trusted_code is not True:
        raise ValueError("Explicitly set TRUSTED_CODE=True only for your own or reviewed code.")
    _check_run_options(("local",), seeds, timeout, rules)
    if not _finite_int(maximum_runs, 1) or 2*len(seeds) > maximum_runs:
        raise ValueError("Own-attack plan exceeds MAX_LOCAL_RUNS; reduce seeds or explicitly increase that local safety guard.")
    with Path(archive).expanduser().open("rb") as stream:
        blob = stream.read(MAX_ARCHIVE+1)
    checked = validate_submission_archive(blob, rules)
    fingerprint = hashlib.sha256(blob).hexdigest()
    print(f"Explicit own-code local execution: {2*len(seeds)} attack instances, timeout {timeout}s PER INSTANCE.")
    with tempfile.TemporaryDirectory(prefix="qduel-own-input-") as tmp:
        fixed = Path(tmp)/"submission.zip"; fixed.write_bytes(blob)
        rows = test_own_attacks(fixed, rules, seeds=tuple(seeds), trusted_code=True, timeout=timeout)
    for row in rows:
        slot = row["slot"]
        row["case"] = f"Your attack {slot}: " + checked["attacks"][slot-1]["name"]
    evidence = {"archive_sha256": fingerprint, "rules_sha256": digest(rules.model_dump()),
                "seeds": list(seeds), "origin": "Existing notebook own-attack helper; explicit local execution",
                "result": {"status": "PASSED" if rows and all(r["status"] == "PASSED" for r in rows) else "FAILED",
                           "cases": rows, "scope": "Local self-test, not a measure of offensive strength against opponents"}}
    return evidence, _save_new_json(output_dir, "own_attack_evidence", evidence)
````

## 7. Decide what to investigate next

Start with the recorded outcome, then use your own logs to identify the cause. The report can separate execution, circuit legality, and recovery. It cannot tell whether an unseen internal estimate was accurate or where your program spent its time.

| What you see | What it establishes | What to investigate |
|---|---|---|
| **Completed; full credit** | All three recorded checkpoints earn full credit on this local case. | Try other permitted structures and seeds to assess the scope of that success. |
| **Completed; ZERO recovery points** | The recorded legal patches all have error at or above the zero-point threshold. | Check the actual patches, your model and measurements, and the conversion from estimate to circuit. Completion alone does not identify which step failed. |
| **Legal identity / empty patch** | The checkpoint recorded no correction gates. | Check whether a patch was submitted, whether a fallback selected identity, and when checkpoints closed. The record alone cannot distinguish these causes. |
| **36 / 36 entanglers** | The recorded patch uses the full entangler allowance. | Inspect synthesis choices and accuracy. Reaching the cap does not by itself explain poor recovery. |
| **Failed; recovery unavailable** | A worker failure was recorded without complete usable checkpoint results. | Read the error and inspect your own logs. Missing scores are not measured zeroes. |
| **Changed, unbound, or different rules** | These results are not established for the selected ZIP and final profile. | Check the paths and profile, rebuild if needed, and test the intended ZIP. |

Compare **raw process errors as well as points**. Errors `0.2` and `0.8` both earn zero points, but they describe different local outcomes. Better error can help you track debugging progress even before it crosses a scoring threshold. It does not earn extra points outside the published scoring rule.

For the open final, ask two separate questions: **Can my method estimate this kind of operation? Can it turn that estimate into an accurate, legal correction before the checkpoint closes?** Then verify which patch actually reached each checkpoint. This notebook supplies that recorded evidence; it supplies no new inference or synthesis algorithm.
